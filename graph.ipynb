```python
import matplotlib.pyplot as plt

years = [2019, 2020, 2021, 2022, 2023, 2024, 2025]
opening_rank = [203, 222, 191, 368, 374, 431, 339]
closing_rank = [616, 584, 521, 608, 674, 656, 673]

plt.figure(figsize=(10, 6))

# Plot lines
plt.plot(years, opening_rank, marker='o', linewidth=2, label='Opening Rank')
plt.plot(years, closing_rank, marker='o', linewidth=2, label='Closing Rank')

# Add data labels
for x, y in zip(years, opening_rank):
    plt.annotate(
        str(y),
        (x, y),
        textcoords="offset points",
        xytext=(0, 8),
        ha='center'
    )

for x, y in zip(years, closing_rank):
    plt.annotate(
        str(y),
        (x, y),
        textcoords="offset points",
        xytext=(0, 8),
        ha='center'
    )

plt.title('JEE Advanced Cutoffs (2019–2025)')
plt.xlabel('Year')
plt.ylabel('Rank')
plt.xticks(years)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()

plt.tight_layout()
plt.show()
```
